## OOPS Tutorial in Python

Classes, Variables, Methods and Objects

## Class

A class is a blueprint for creating objects.
It defines properties (variables) and behaviors (methods).

class Student:
    pass

👉 Think of it like a design of a house (not the actual house).

## 2. Object

An object is an instance of a class.
It is a real-world entity created using the class.

s1 = Student()

👉 Here, s1 is an object of Student.

## 3. Variables (Attributes)

Variables inside a class are called attributes.
They store data for objects.

Types:
🔹 Instance Variables (unique for each object)
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

## 4. Methods

Methods are functions inside a class that define behavior.

class Student:

```python
class Car:
    def __init__(self,window,door,enginetype):
        self.windows=window
        self.doors=door
        self.enginetype=enginetype
    def self_driving(self):
        return "This is a {} car".format(self.enginetype)
```

```python
car1=Car(4,5,"gas")
```

```python
car1.self_driving()
```

```python
car2=Car(3,4,"diesal")
```
