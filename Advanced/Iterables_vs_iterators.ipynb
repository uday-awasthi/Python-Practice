## Python List Iterables vs Iterators

## Iterable

#### An iterable is an object that can be looped over. It provides an __iter__() method that returns an iterator.

Examples of iterables:

my_list = [1, 2, 3]

my_string = "hello"

my_tuple = (1, 2, 3)

my_dict = {"a": 1, "b": 2}

You can use them directly in a for loop:

for item in my_list:
    print(item)

```python
## List is Iterable

lst = [1,2,3,4,5,6,7]

for i in lst:
    print(i)
```

```python
lst1=iter(lst)
```

```python
lst1
```

```python
next(lst1)
```

```python
for i in lst1:
    print(i)
```

## Iterator

#### An iterator is an object that keeps track of its current position during iteration.

It must implement:

__iter__() → returns itself
__next__() → returns the next item

Example:

my_list = [1, 2, 3]

iterator = iter(my_list)

print(next(iterator))  # 1
print(next(iterator))  # 2
print(next(iterator))  # 3

Output:

1
2
3

Calling next() again raises:

StopIteration

## Relationship
iterable --> iter() --> iterator --> next()

Example:

numbers = [10, 20, 30]   # iterable

it = iter(numbers)       # iterator

print(next(it))          # 10
print(next(it))          # 20
print(next(it))          # 30
Checking Types
from collections.abc import Iterable, Iterator

nums = [1, 2, 3]

print(isinstance(nums, Iterable))  # True
print(isinstance(nums, Iterator))  # False

it = iter(nums)

print(isinstance(it, Iterable))    # True
print(isinstance(it, Iterator))    # True

Output:

True
False
True
True
