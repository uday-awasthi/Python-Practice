## Questions

### What is string formatting in Python?

String formatting is the process of inserting variables, values, or expressions into a string to create a readable and meaningful output.

Example:

name = "Ali"
print(f"Hello, {name}")

Output:

Hello, Ali


### 2. Why is string formatting used?

String formatting is used to:

Insert variables into strings.
Create dynamic output.
Improve code readability.
Control the appearance of numbers, dates, and text.

Example:

name = "Sara"
age = 20

print(f"My name is {name} and I am {age} years old.")

### 3. What are the different ways to format strings in Python?

Python provides three main ways:

a.  % Formatting (Old Style)
name = "Ali"
print("Hello %s" % name)

b. str.format() Method

name = "Ali"
print("Hello {}".format(name))

c. f-Strings (Modern Method)
name = "Ali"
print(f"Hello {name}")

#### 4. What is an f-string?

An f-string (formatted string literal) is a string prefixed with f that allows variables and expressions to be embedded directly inside curly braces {}.

Example:

name = "John"
age = 25

print(f"My name is {name} and I am {age} years old.")

Output:

My name is John and I am 25 years old.

You can also use expressions:

x = 5
y = 10

print(f"Sum = {x + y}")

Output:

Sum = 15

#### 5. Which string formatting method is preferred in modern Python?

f-strings are the preferred method in modern Python (Python 3.6+).

Reasons:

Easier to read
Faster than other methods
Cleaner syntax
Supports expressions directly

Example:

name = "Ali"
marks = 95

print(f"{name} scored {marks} marks.")

Output:

Ali scored 95 marks.
