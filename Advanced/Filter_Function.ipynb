## Filter Function in Python

The filter() function is used to select elements from an iterable based on a condition.

It returns only those elements for which the function returns True.

Syntax:

filter(function, iterable)

1. function → A function that returns True or False

2 .iterable → A list, tuple, set, etc.

```python
def even(num):
    if num % 2 == 0:
        return True
```

```python
lst =[1,2,3,4,5,6,7,8,9,0]
```

```python
list(filter(even,lst))
```

```python
list(filter(lambda num:num%2==0,lst))
```

```python
list(map(lambda num:num%2==0,lst))
```

## Filter Positive Numbers

```python
numbers = [-5, -2, 0, 3, 7, -1]

result = list(filter(lambda x: x > 0, numbers))

print(result)
```

## Filter Names by Length

```python
names = ['Uday', 'Rahul', 'Aman', 'Priya']

result = list(filter(lambda x: len(x) > 4, names))

print(result)
```

## Filter Names by Length

```python
data = ['Python', '', 'Pandas', '', 'NumPy']

result = list(filter(None, data))

print(result)
```

## Filter Missing Values

```python
data = [10, None, 20, None, 30]

clean_data = list(filter(lambda x: x is not None, data))

print(clean_data)
```

## Filter Missing Values

```python
salaries = [25000, 50000, 75000, 100000]

high_salary = list(filter(lambda x: x >= 50000, salaries))
```
