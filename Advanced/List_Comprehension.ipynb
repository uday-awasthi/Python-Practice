## List Comprhension

List  comprehesion provide a concide way to crete lists. It consists of brackets containing an expression followed by a for clause, then zero or more for or if cluses. The expression can be anything, meaning you can put in all kinds of objects in lists.

```python
lst1 = []
def lst_square(lst):
    for i in lst:
        lst1.append(i*i)
    return lst1
```

```python
lst_square([1,2,3,4,5,6,7])
```

```python
lst=[1,2,3,4,5,6,7]
```

```python
## List comprehension
lst1=[i*i for i in lst if i%2!=0]
```

```python
print(lst1)
```
