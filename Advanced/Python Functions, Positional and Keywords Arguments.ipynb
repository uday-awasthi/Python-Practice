## Positional Arguments

When arguments are passed to a function in the same order as the parameters are defined, they are called positional arguments.


## Keyword Arguments

When arguments are passed using the parameter names, they are called keyword arguments.

##### Example
def introduce(name, age):
    print("Name:", name)
    print("Age:", age)

introduce(age=20, name="Uday")

#### Output:

Name: Uday
Age: 20


# Positional arguments
print(add(10, 20))

# Keyword arguments
print(add(a=10, b=20))
print(add(b=20, a=10))

Output:

30
30

```python
## positional argument
## keyword argument

def hello(name,age=29):
    print("My name is {} and age is {}".format(name,age))
```

```python
hello('Uday')
```

```python
def hello(*args,**kwargs):
    print(args)
    print(kwargs)
```

```python
hello("Uday","Naik",age=29,dob= 2007)
```

```python
lst=['krish','Naik']
dict_args={'age': 29, 'dob' : 1990}
```

```python
hello(*lst,**dict_args)
```

```python
lst=[1,2,3,4,5,6,7,8]
```

```python
def evenoddsum(lst):
    even_sum=0
    odd_sum=0
    for i in lst:
        if i %2==0:
            even_sum=even_sum+i
        else:
            odd_sum=odd_sum+i
    return even_sum,odd_sum
```

```python
evenoddsum(lst)
```

 ## Combining Positional and Keyword Arguments

You can use both together, but positional arguments must come first.

Example:

def student(name, age, city):
    print(name, age, city)

student("Uday", age=20, city="Lucknow")
