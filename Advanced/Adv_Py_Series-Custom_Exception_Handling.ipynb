```python
year = int(input("Enter the year of Birth"))
age = 2026 - year

try:
    if age <= 30 & age>20:
        print("The age is valid. You can apply for the exams")
    else:
        raise dobException()

except dobException:
    print("The  age is not within the range. You cannot apply for the age")
```
