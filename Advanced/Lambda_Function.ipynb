## Lambda Function

anonymous function.

A function with no name


```python
def addition(a,b):
    return a+b
```

```python
addition(4,5)
```

```python
addition = lambda a,b:a+b
```

```python
 addition(9,6)
```

```python
def even(num):
    if num % 2==0:
        return True
```

```python
even(24)
```

```python
even1=lambda a:a%2==0
```

```python
even1(12)
```

```python
def addition(x,y,z):
    return x+y+z
```

```python
addition(2,3,4)
```

```python
addition = lambda x,y,z: x+y+z
```

```python
addition(4,5,6)
```

```python
## Using With map()
nums = [1,2,3,4,5]

squares = list(map(lambda X: X**2, nums))

print(squares)
```

```python
## Using with filter()

nums = [1, 2, 3, 4, 5, 6]

evens = list(filter(lambda x : x % 2 == 0, nums))

print(evens)
```

```python
## Using with sorted()

students = [
    ("Uday", 85),
    ("Rahul", 92),
    ("Aman", 78)
]

sorted_students = sorted(students, key=lambda x:x[1])

print(sorted_students)
```
