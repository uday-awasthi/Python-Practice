## Map Function In Python :

map() is a built-in Python function that applies a specified function to every element of an iterable and returns an iterator containing the transformed values.

Syntax :

map(function, iterable)

```python
def even_or_odd(num):
    if num%2==0:
        return "The number  {} is Even".format(num)
    else:
        return "The number {} id Odd".format(num)
```

```python
even_or_odd(24)
```

```python
lst=[1,2,3,4,5,6,7,8,9,24,54,67,89,98]
```

```python
list(map(even_or_odd,lst))
```

## Convert Strings to Integers

```python
nums = ['1', '2', '3', '4']

result = list(map(int, nums))

print(result)
```

## Convert Names to Uppercase

```python
names = ['uday', 'rahul', 'aman']

result = list(map(str.upper, names))

print(result)
```

## Multiple Iterables

```python
a = [1, 2, 3]
b = [4, 5, 6]

result = list(map(lambda x, y: x + y, a, b))

print(result)
```
