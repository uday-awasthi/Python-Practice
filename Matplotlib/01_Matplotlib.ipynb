## MatplotLib Tutorial
Matplotlib is a plotting library for the Python programming language and its numerical mathematics extension NumPy. It provides an object-oriented API for embedding plots into applications using general-purpose GUI toolkits.

Some of the major Pros of Matplotlib are:
- Generally easy to get started for simple plots
- Support for custom labels and texts
- Great control of every element in figure
- High-quality outputs in many formats
- Very customizable in general

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
import numpy as np
```

```python
## Simple Examples
x = np.arange(0,10)
y = np.arange(11,21)
```

```python
a = np.arange(40,50)
b = np.arange(50,60)
```

## plt.scatter
Definition: Creates a scatter plot to show the relationship between two numerical variables using individual data points.

```python
plt.scatter(x, y, c='g')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('Graph in 2D')
plt.show()
```

```python
y = x * x
```

## plt.plot
Definition: Creates a line plot used to visualize trends, patterns, or changes in data over a continuous interval.

```python
plt.plot(x, y, 'r*', linestyle='dashed', linewidth=2, markersize=12)
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('2d Diagram')
plt.show()
```

## Creating Subplots
Definition: Subplots allow multiple plots to be displayed within a single figure for easy comparison and analysis.

```python
plt.figure(figsize=(10, 8))
plt.subplot(2, 2, 1)
plt.plot(x, y, 'r*--')

plt.subplot(2, 2, 2)
plt.plot(x, y, 'g*-')

plt.subplot(2, 2, 3)
plt.plot(x, y, 'bo')

plt.subplot(2, 2, 4)
plt.plot(x, y, 'go')

plt.tight_layout()
plt.show()
```

```python
x = np.arange(1, 11)
y = 3 * x + 5
plt.title("Matplotlib demo")
plt.xlabel("x axis caption")
plt.ylabel("y axis caption")
plt.plot(x, y)
plt.show()
```

## Sine Wave

```python
x = np.arange(0, 4 * np.pi, 0.1)
y = np.sin(x)
plt.title("Sine wave form")
plt.plot(x, y)
plt.show()
```

## Sine and Cosine

```python
x = np.arange(0, 5 * np.pi, 0.1)
y_sin = np.sin(x)
y_cos = np.cos(x)

plt.subplot(2, 1, 1)
plt.plot(x, y_sin, 'r--')
plt.title('Sine')

plt.subplot(2, 1, 2)
plt.plot(x, y_cos, 'g--')
plt.title('Cosine')

plt.tight_layout()
plt.show()
```

## Bar Plot
A bar plot represents categorical data using rectangular bars whose heights or lengths correspond to the values.

```python
x = [2, 8, 10]
y = [11, 16, 9]
x2 = [3, 9, 11]
y2 = [6, 15, 7]

plt.bar(x, y, align='center', label='Bar 1')
plt.bar(x2, y2, color='g', label='Bar 2')
plt.title('Bar graph')
plt.ylabel('Y axis')
plt.xlabel('X axis')
plt.legend()
plt.show()
```

## Histograms
A histogram displays the frequency distribution of numerical data by dividing it into intervals called bins.

```python
a = np.array([22, 87, 5, 43, 56, 73, 55, 54, 11, 20, 51, 5, 79, 31, 27])
plt.hist(a, bins=30)
plt.title("Histogram")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

## Box Plot
A box plot summarizes data distribution using the minimum, first quartile, median, third quartile, and maximum values, helping identify outliers.

```python
data = [np.random.normal(0, std, 100) for std in range(1, 4)]
plt.boxplot(data, vert=False, patch_artist=True)
plt.title("Box Plot")
plt.show()
```

## Pie Chart
A pie chart is a circular chart divided into slices, where each slice represents a category's proportion of the whole dataset.

```python
labels = ['Python', 'C++', 'Ruby', 'Java']
sizes = [213, 130, 245, 210]
colors = ['gold', 'yellowgreen', 'lightcoral', 'blue']
explode = (0.1, 0, 0, 0)

plt.pie(sizes, explode=explode, labels=labels, colors=colors, autopct='%1.1f%%', shadow=True)
plt.title("Programming Languages")
plt.axis('equal')
plt.show()
```
